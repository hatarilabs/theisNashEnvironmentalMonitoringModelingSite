```python
import geopandas as gpd
from shapely.geometry import Polygon, MultiPolygon


# 1. Load the multipolygon shapefile
gdf = gpd.read_file('../Shp/River_Channel_2023.shp')

# 2. "Explode" the multipolygons into individual polygon rows
# This turns a MultiPolygon with 3 parts into 3 separate Polygon rows
gdf_single = gdf.explode(index_parts=False)

# 3. Double check: Ensure all geometries are Polygons
# If any were still MultiPolygons (rare after explode), this filters them
gdf_single = gdf_single[gdf_single.geometry.type == 'Polygon']

output_schema = {
    'geometry': 'Polygon',
    'properties': {col: 'str' for col in gdf_single.columns if col != 'geometry'}
}

# 4. Save to a new shapefile
# We explicitly set the schema to 'Polygon' if necessary
gdf_single.to_file('../Shp/River_Channel_2023_spath1.shp', 
                   driver='ESRI Shapefile',
                   schema=output_schema,
                   engine='fiona')
    
#print(f"Successfully converted {input_path} to single-part polygons at {output_path}")
```
